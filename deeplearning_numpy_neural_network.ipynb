```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
prac_data=pd.read_csv('/content/drive/MyDrive/Machine learning files/datasets/messy_medical_data.csv')
prac_data
```

```python
prac_data['Sex'].unique()
```

```python
prac_data['Sex']=prac_data['Sex'].str.replace('M','male')
prac_data['Sex']=prac_data['Sex'].str.replace('F','female')
prac_data['Sex']=prac_data['Sex'].str.replace('MALE','male')
prac_data['Sex']=prac_data['Sex'].map({'female':1,'male':0})
prac_data['Sex']=prac_data['Sex'].astype(float)
```

```python
prac_data['Smoker'].unique()
```

```python
prac_data['Smoker']=prac_data['Smoker'].str.replace('no','NO')
prac_data['Smoker']=prac_data['Smoker'].str.replace('No','NO')
prac_data['Smoker']=prac_data['Smoker'].map({'YES':1,'NO':0})
```

```python
prac_data=prac_data.drop('Region',axis=1)
prac_data=prac_data.drop('BP',axis=1)
```

```python
prac_data['Cholesterol']=prac_data['Cholesterol'].str.replace('200mg','200').astype(float)
prac_data['Charges']=prac_data['Charges'].str.replace('$16884.92','16884.92').astype(float)
```

```python
x_train,x_temp,y_train,y_temp=train_test_split(x,y,test_size=0.2,random_state=42)
x_dev,x_test,y_dev,y_test=train_test_split(x_temp,y_temp,test_size=0.5,random_state=42)
```

```python
scaler1=StandardScaler()
x_train_scaled=scaler1.fit_transform(x_train)
x_dev_scaled=scaler1.transform(x_dev)
x_test_scaled=scaler1.transform(x_test)
```

```python
scaler2=StandardScaler()
y_train_scaled=scaler2.fit_transform(y_train)
y_dev_scaled=scaler2.transform(y_dev)
y_test_scaled=scaler2.transform(y_test)
```

```python
x_train_scaled.shape
```

```python
#np.random.seed(0)

layer1_w=np.random.randn(10,6)*np.sqrt(2.0/x_train_scaled.shape[1])  #first layer weights
layer1_b=np.zeros((1,10))   #first layer bias

layer2_w=np.random.randn(20,10)*np.sqrt(2.0/x_train_scaled.shape[1]) #second layer weights
layer2_b=np.zeros((1,20))  #second layer bias

layer3_w=np.random.randn(10,20)*np.sqrt(2.0/x_train_scaled.shape[1]) #layer 3 weights
layer3_b=np.zeros((1,10)) # third layer bias

output_w=np.random.randn(1,10)*np.sqrt(2.0/x_train_scaled.shape[1])   #output neuron weights
output_b=np.zeros((1,1))  #output neuron bias

learning_rate=0.001  #learning rate

batch_size=x_train_scaled.shape[0] #batch size

reg_lambda=0.001
```

```python
np.random.seed(42)

for epoch in range(1000):

  z1=np.dot(x_train_scaled,layer1_w.T)+layer1_b #foward pass
  a1=np.maximum(0,z1)  #activation function
  relu_grad1=(z1>0).astype(float)  #relu gradient
  layer1_norm2=np.linalg.norm(layer1_w,ord=2)  #layer 1 norm


  z2=np.dot(a1,layer2_w.T)+layer2_b #foward pass
  a2=np.maximum(0,z2)  #activation function
  relu_grad2=(z2>0).astype(float) #relu gradinet
  layer2_norm2=np.linalg.norm(layer2_w,ord=2) #layer 2 norm


  z3=np.dot(a2,layer3_w.T)+layer3_b  #foward pass
  a3=np.maximum(0,z3)   #activation function
  relu_grad3=(z3>0).astype(float)  #relu gradient
  layer3_norm2=np.linalg.norm(layer3_w,ord=2)  #layer 3 norm


  z4=np.dot(a3,output_w.T)+output_b #forwad pass
  a4=z4  #linear activation
  output_norm=np.linalg.norm(output_w,ord=2) #output neuron norm

  loss=0.5*(y_train_scaled-a4)**2  #losss function
  l2_penalty=(reg_lambda)/(2*batch_size)*(np.sum(layer1_w**2)+np.sum(layer2_w**2)+np.sum(layer3_w**2)+np.sum(output_w**2))
  loss=loss+l2_penalty

  #erros
  delta_output=a4-y_train_scaled  #output neuron error
  delta_layer3=np.dot(delta_output,output_w)*relu_grad3  #layer3 error
  delta_layer2=np.dot(delta_layer3,layer3_w)*relu_grad2  #layer2 error
  delta_layer1=np.dot(delta_layer2,layer2_w)*relu_grad1  #layer1 error


  #bias gradient
  output_b_grad=np.mean(delta_output,axis=0,keepdims=True)
  layer3_b_grad=np.mean(delta_layer3,axis=0,keepdims=True)
  layer2_b_grad=np.mean(delta_layer2,axis=0,keepdims=True)
  layer1_b_grad=np.mean(delta_layer1,axis=0,keepdims=True)

  #weight gradient
  output_w_grad=np.dot(delta_output.T,a3)/batch_size + (reg_lambda/batch_size)*output_w
  layer3_w_grad=np.dot(delta_layer3.T,a2)/batch_size + (reg_lambda/batch_size)*layer3_w
  layer2_w_grad=np.dot(delta_layer2.T,a1)/batch_size + (reg_lambda/batch_size)*layer2_w
  layer1_w_grad=np.dot(delta_layer1.T,x_train_scaled)/batch_size+(reg_lambda/batch_size)*layer1_w

  #update
  #output neuron
  output_w=output_w-learning_rate*output_w_grad
  output_b=output_b-learning_rate*output_b_grad

  #layer3
  layer3_w=layer3_w-learning_rate*layer3_w_grad
  layer3_b=layer3_b-learning_rate*layer3_b_grad

  #layer2
  layer2_w=layer2_w-learning_rate*layer2_w_grad
  layer2_b=layer2_b-learning_rate*layer2_b_grad

  #layer1
  layer1_w=layer1_w-learning_rate*layer1_w_grad
  layer1_b=layer1_b-learning_rate*layer1_b_grad


  if epoch %100==0:
    print(f'epoch:{epoch}, loss:{loss.mean().item():.2f},  norm_layer1:{layer1_norm2},  norm_layer2:{layer2_norm2}, norm_layer3:{layer3_norm2}, output_neuron:{output_norm}')







```
